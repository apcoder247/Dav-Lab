```python
def arith_opr():
    a=int(input("Enter the first number: "))
    b=int(input("\nEnter the second number: "))
    choice=input("Choose your desired operation: ")
    if choice=="+":
        result=a+b
    elif choice=="-":
        result=a-b
    elif choice=="*":
        result=a*b
    elif choice=="/":
        if b!=0:
            result=a/b
        else:
            print("DIVISION BY ZERO NO ALLOWED")
            return
    else:
        print("Invalid input")
        return
    print("The required answer is: ",result)
arith_opr()
        
    
    
```

```python
def prime(n):
    flag=0
    if n<=1:
        return False
    elif n==2:
        return True
    for i in range(2,int(n**0.5)+1):
        if n%i==0:
            flag=1
            break
    if flag==0:
        return True
    else:
        return False
num=int(input("Enter the number: "))
print(prime(num))
```

```python
def sequence():
    sum=0
    print("Odd numbers are:\n")
    for i in range(1,51):
        print(2*i-1,end=" ")
        sum+=2*i
    print()
    print("Sum of all even numbers is: ",sum)
sequence()
    
```

```python
def add_elements():
    t = input("Enter type (int/float/string): ")

    if t == "int":
        x = int(input("Enter first number: "))
        y = int(input("Enter second number: "))
        print("Result:", x + y)

    elif t == "float":
        x = float(input("Enter first number: "))
        y = float(input("Enter second number: "))
        print("Result:", x + y)

    elif t == "string":
        x = input("Enter first string: ")
        y = input("Enter second string: ")
        print("Result:", x + y)

    else:
        print("Invalid type entered!")

add_elements()
```

```python
def content():
    str=input("Enter your string: ")
    count_v=0
    count_c=0
    for i in str:
        if i.isalpha():
            if i in "AEIOUaeiou":
                count_v+=1
            else:
                count_c+=1
    print("Number of vowels are ",count_v," and number of consonants are ",count_c)
content()
    
```
